```python
delete the deck column becaue the missing value percentage of check column is 77 that is greater than the 20% which exceeds condition
age column have < 20% of missing value percentage , we can fill the posistion with median
embarked column have < 20% of loss value percentage , we can fill the posistion with median/mode(mode if the columns is non-numerical)
embarked_town column have < 20 % of loss value percentage , we can fill the posistion with median/mode
```

```python
da.dropna()
```

```python
da.drop('deck',axis=1,inplace=True)
```

```python
da.columns
```

```python
da['age'].fillna(da['age'].median(),inplace=True)
```

```python
da['embark_town'].fillna(da['embark_town'].mode()[0],inplace=True)
```

```python
da['embarked'].fillna(da['embarked'].mode()[0],inplace=True)
```
